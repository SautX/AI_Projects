```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid")
df = pd.read_csv("/content/loan_data.csv", sep=None, engine='python', on_bad_lines='skip')
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="TARGET", data=df, palette="Set2", hue="TARGET", legend=False)
plt.title("Loan Status Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["AMT_CREDIT"], bins=30, kde=True, color="purple")
plt.title("Loan Amount Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="TARGET", y="AMT_CREDIT", data=df, palette="coolwarm", hue="TARGET", legend=False)
plt.title("Loan Amount vs Loan Status")
plt.show()
```

```python
df["DTI"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.histplot(df["AMT_INCOME_TOTAL"], bins=40, kde=True, color="green")
plt.title("Annual Income Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="TARGET", y="AMT_INCOME_TOTAL", data=df, palette="viridis")
plt.title("Annual Income vs Default")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["DTI"], bins=30, kde=True, color="red")
plt.title("Debt-to-Income Ratio Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="TARGET", y="DTI", data=df, palette="plasma")
plt.title("Debt-to-Income Ratio vs Default")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="TARGET", y="AMT_CREDIT", data=df, palette="coolwarm")
plt.title("Credit Amount vs Default")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="TARGET", y="AMT_ANNUITY", data=df, palette="Spectral")
plt.title("Loan Annuity vs Default")
plt.show()
```

```python
plt.figure(figsize=(10,6))

# Convert specified columns to numeric, coercing errors to NaN
for col in ["TARGET", "AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY", "DTI"]:
    df[col] = pd.to_numeric(df[col], errors='coerce')

sns.heatmap(
    df[["TARGET","AMT_INCOME_TOTAL","AMT_CREDIT","AMT_ANNUITY","DTI"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```

```python
dti_default = df.groupby(pd.cut(df["DTI"], bins=20))["TARGET"].mean()

plt.figure(figsize=(8,4))
plt.plot(dti_default.values, marker='o', color="crimson")
plt.title("Default Rate vs Debt-to-Income Ratio")
plt.xlabel("DTI Buckets")
plt.ylabel("Default Probability")
plt.show()
```
